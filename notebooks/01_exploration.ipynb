# CreditWatch -- Initial Data Exploration

This notebook inspects the raw UCI Polish Companies Bankruptcy Dataset (5year.arff) BEFORE any cleaning or feature engineering decisions are hardcoded into `src/`. Run this after placing `5year.arff` in `data/raw/`.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
sys.path.insert(0, os.path.abspath('../src'))

from data_loader import load_raw_dataset
from config import TARGET_COLUMN

df = load_raw_dataset()
print(df.shape)
df.head()
```

```python
# Target distribution -- confirms the class imbalance discussed in the README
df[TARGET_COLUMN].value_counts(normalize=True)
```

```python
# Missingness per column -- informs the MAX_MISSING_FRACTION threshold in preprocessing.py
df.drop(columns=[TARGET_COLUMN]).isna().mean().sort_values(ascending=False).head(20)
```

```python
# Basic descriptive statistics -- look for infinite values / extreme outliers
import numpy as np
numeric_df = df.drop(columns=[TARGET_COLUMN])
print('Any +/- inf values present:', np.isinf(numeric_df.values).any())
numeric_df.describe().T.head(20)
```
